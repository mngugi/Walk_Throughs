```python
import random
import csv

with open("transactions.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["amount"])

    for _ in range(5000):
        exponent = random.uniform(2, 6)
        amount = round(10 ** exponent, 2)
        writer.writerow([amount])

print("transactions.csv created")
```
