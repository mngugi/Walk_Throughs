```python
import pandas as pd
import numpy as np
from collections import Counter

data = pd.read_csv("transactions.csv")

first_digits = [
    int(str(abs(x))[0])
    for x in data["amount"]
    if x > 0
]

counts = Counter(first_digits)
total = len(first_digits)

for d in range(1, 10):
    observed = counts[d] / total
    expected = np.log10(1 + 1/d)
    print(d, f'Observed: {observed},Expected {expected}')
```
